```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
area = np.random.randint(500, 5001, 100)
print(area)
```

```python
bedrooms = np.random.randint(1, 7, 100)
print(bedrooms)
```

```python
price = 2000000 + (area * 5000) + (bedrooms * 500000)
print(price)
```

```python
df = pd.DataFrame({"Area": area, "Bedrooms": bedrooms, "Price": price})
print(df.head())
```

```python
df.to_csv("housing.csv", index=False)
```

```python
print(df.shape)
```

```python
print(df.isnull().sum())
```

```python
X = df[["Area", "Bedrooms"]]
y = df["Price"]

print(X.head())  # dataframe hence double brackets
print(y.head())  # series hence single bracket
```

```python
X_Train = X.iloc[:80]
X_Test = X.iloc[80:]

y_Train = y.iloc[:80]
y_Test = y.iloc[80:]

print(X_Train.shape)
print(X_Test.shape)
print(y_Train.shape)
print(y_Test.shape)
```

```python
X_Train = X_Train.to_numpy()
X_Test = X_Test.to_numpy()

y_Train = y_Train.to_numpy()
y_Test = y_Test.to_numpy()

print(X_Train.shape)
print(type(X_Train))
```

```python
weights = np.zeros(2)
bias = 0
print(weights)
print(bias)

predictions = np.dot(X_Train, weights) + bias

errors = predictions - y_Train
print(errors[:5])
```

```python
mse = np.mean(errors**2)
print(mse)
```

```python
learning_rate = 0.01
```

```python
X_Train_scaled = X_Train.astype(float)
X_Train_scaled[:, 0] = X_Train_scaled[:, 0] / 5000

y_Train_scaled = y_Train / 10000000

print(X_Train_scaled[:5])
print(y_Train_scaled[:5])
```

```python
weights = np.zeros(2)
bias = 0

for i in range(10000):
    predictions = np.dot(X_Train_scaled, weights) + bias
    errors = predictions - y_Train_scaled

    weight_gradients = 2 * np.dot(X_Train_scaled.T, errors) / len(y_Train_scaled)
    bias_gradient = 2 * np.mean(errors)

    weights = weights - learning_rate * weight_gradients
    bias = bias - learning_rate * bias_gradient

print("Trained weights:", weights)
print("Trained bias:", bias)
```

```python
X_Test_scaled = X_Test.astype(float)
X_Test_scaled[:, 0] = X_Test_scaled[:, 0] / 5000

print(X_Test_scaled[:5])

X_predictions_scaled = np.dot(X_Test_scaled, weights) + bias
X_predictions = X_predictions_scaled * 10000000

print(X_predictions[:5])
```

```python
test_error = X_predictions - y_Test
test_mse = np.mean(test_error**2)

print("Testing mse: ", test_mse)

rmse = np.sqrt(test_mse)
print("Actual mse: ", rmse)
```
